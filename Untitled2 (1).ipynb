```python
import random
import os
import pandas as pd

# Fix seed
random.seed(42)

all_files = os.listdir(path)
sample_files = random.sample(all_files, 10000)

data = []

for file in sample_files:
    with open(path + "/" + file, "r") as f:
        lines = f.readlines()

        props = lines[1].split()

        # HOMO-LUMO GAP (index 8)
        gap = float(props[8])

        elements = [line.split()[0] for line in lines[2:]]

        count_C = elements.count("C")
        count_N = elements.count("N")
        count_O = elements.count("O")
        count_F = elements.count("F")
        heavy_atoms = sum(1 for e in elements if e != "H")

        data.append([gap, heavy_atoms, count_C, count_N, count_O, count_F])

df = pd.DataFrame(
    data,
    columns=["gap", "heavy_atoms", "C", "N", "O", "F"]
)

print("New dataset shape:", df.shape)
df.head()
```

```python
from sklearn.model_selection import train_test_split

df_train_pool, df_test = train_test_split(
    df, test_size=0.2, random_state=42
)

print("Train pool size:", df_train_pool.shape)
print("Test set size:", df_test.shape)

```

```python
import numpy as np

# Strategy 1 — Random sampling (unchanged)
def sample_random(df_pool, n, seed):
    return df_pool.sample(n=n, random_state=seed)


# Strategy 2 — Stratified by molecular size
def sample_stratified_size(df_pool, n, seed):
    np.random.seed(seed)

    groups = df_pool.groupby("heavy_atoms")
    samples = []

    for _, group in groups:
        proportion = len(group) / len(df_pool)
        n_group = int(round(n * proportion))

        if n_group > 0:
            n_group = min(n_group, len(group))
            samples.append(group.sample(n=n_group, random_state=seed))

    df_sample = pd.concat(samples)

    # Adjust if rounding mismatch
    if len(df_sample) > n:
        df_sample = df_sample.sample(n=n, random_state=seed)
    elif len(df_sample) < n:
        remaining = df_pool.drop(df_sample.index)
        extra = remaining.sample(n=n - len(df_sample), random_state=seed)
        df_sample = pd.concat([df_sample, extra])

    return df_sample


# Strategy 3 — Target-binned stratified
def sample_stratified_target(df_pool, n, seed, bins=5):
    np.random.seed(seed)

    df_copy = df_pool.copy()
    df_copy["gap_bin"] = pd.qcut(df_copy["gap"], q=bins, duplicates="drop")

    groups = df_copy.groupby("gap_bin")

    samples = []

    for _, group in groups:
        proportion = len(group) / len(df_copy)
        n_group = int(round(n * proportion))

        if n_group > 0:
            n_group = min(n_group, len(group))
            samples.append(group.sample(n=n_group, random_state=seed))

    df_sample = pd.concat(samples)

    # Adjust if rounding mismatch
    if len(df_sample) > n:
        df_sample = df_sample.sample(n=n, random_state=seed)
    elif len(df_sample) < n:
        remaining = df_copy.drop(df_sample.index)
        extra = remaining.sample(n=n - len(df_sample), random_state=seed)
        df_sample = pd.concat([df_sample, extra])

    return df_sample.drop(columns=["gap_bin"])

```

```python
from sklearn.kernel_ridge import KernelRidge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error


def train_and_evaluate(train_df, test_df):
    feature_cols = ["heavy_atoms", "C", "N", "O", "F"]

    X_train = train_df[feature_cols].values
    y_train = train_df["gap"].values

    X_test = test_df[feature_cols].values
    y_test = test_df["gap"].values

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("krr", KernelRidge(kernel="rbf", alpha=1.0, gamma=0.1))
    ])

    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    return mean_absolute_error(y_test, preds)
```

```python
train_sizes = [50, 100, 200, 500, 1000, 2000, 5000]
n_seeds = 5


```

```python
# Example single experiment

seed = 0
n = 200

train_sample = sample_random(df_train_pool, n=n, seed=seed)

mae = train_and_evaluate(train_sample, df_test)

print("MAE (Random, n=200, seed=0):", mae)
```

```python
results = []

for n in train_sizes:
    for seed in range(n_seeds):

        train_sample = sample_random(df_train_pool, n=n, seed=seed)
        mae = train_and_evaluate(train_sample, df_test)

        results.append({
            "strategy": "random",
            "n_train": n,
            "seed": seed,
            "mae": mae
        })

print("Random experiments done.")
```

```python
import pandas as pd

df_results_random = pd.DataFrame(results)
df_results_random.head()
```

```python
results = []

for n in train_sizes:
    for seed in range(n_seeds):

        # Random
        train_random = sample_random(df_train_pool, n=n, seed=seed)
        mae_random = train_and_evaluate(train_random, df_test)

        results.append({
            "strategy": "random",
            "n_train": n,
            "seed": seed,
            "mae": mae_random
        })

        # Stratified by size
        train_size = sample_stratified_size(df_train_pool, n=n, seed=seed)
        mae_size = train_and_evaluate(train_size, df_test)

        results.append({
            "strategy": "stratified_size",
            "n_train": n,
            "seed": seed,
            "mae": mae_size
        })

        # Target-binned
        train_target = sample_stratified_target(df_train_pool, n=n, seed=seed)
        mae_target = train_and_evaluate(train_target, df_test)

        results.append({
            "strategy": "stratified_target",
            "n_train": n,
            "seed": seed,
            "mae": mae_target
        })

print("All strategies done.")
```

```python
df_results = pd.DataFrame(results)

summary = (
    df_results
    .groupby(["strategy", "n_train"])["mae"]
    .mean()
    .reset_index()
)

summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

for strategy in summary["strategy"].unique():
    subset = summary[summary["strategy"] == strategy]
    plt.plot(subset["n_train"], subset["mae"], marker="o", label=strategy)

plt.xlabel("Training Set Size")
plt.ylabel("Mean Absolute Error (MAE)")
plt.title("Learning Curves: Sampling Strategy Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

```python
summary_stats = (
    df_results
    .groupby(["strategy", "n_train"])["mae"]
    .agg(["mean", "std"])
    .reset_index()
)

summary_stats
```

```python
plt.figure(figsize=(8,5))

for strategy in summary_stats["strategy"].unique():
    subset = summary_stats[summary_stats["strategy"] == strategy]
    plt.errorbar(
        subset["n_train"],
        subset["mean"],
        yerr=subset["std"],
        marker="o",
        capsize=4,
        label=strategy
    )

plt.xlabel("Training Set Size")
plt.ylabel("Mean Absolute Error (MAE)")
plt.title("Learning Curves with Uncertainty")
plt.legend()
plt.grid(True)
plt.show()
```
